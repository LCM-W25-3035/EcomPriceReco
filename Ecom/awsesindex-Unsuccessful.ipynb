```python
# Function to index data into OpenSearch Serverless
def index_data_to_opensearch(data,collection):
    try:
        response = opensearch_client.index(
            index=collection,
            body=data,
            refresh=True
        )
        print(response)
        print(f"Document indexed: {response['_id']}")
    except Exception as e:
        print(f"Error indexing document: {e}")
```
