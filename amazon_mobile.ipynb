```python
# Import required modules
import requests
from bs4 import BeautifulSoup
```

```python
# Define headers to mimic a browser request
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.5735.90 Safari/537.36"
}
```

```python
# Specify the Amazon product page URL
url = "https://www.amazon.ca/s?k=mobile+phone&crid=193091PYP4O7V&sprefix=mob%2Caps%2C90&ref=nb_sb_ss_ts-doa-p_1_3"
```

```python
# Make a GET request to the URL
response = requests.get(url, headers=headers)
```

```python
# Check if the request was successful
if response.status_code == 200:
    # Parse the page content
    soup = BeautifulSoup(response.content, "html.parser")

    # Extract desired data
    title = soup.find("span", id="productTitle")
    price = soup.find("span", class_="a-offscreen")

    # Display the extracted data
    print("Product Title:", title.get_text(strip=True) if title else "Title not found")
    print("Price:", price.get_text(strip=True) if price else "Price not found")
else:
    print(f"Failed to fetch page. Status code: {response.status_code}")
```
