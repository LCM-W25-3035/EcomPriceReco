```python
# Importing necessary libraries
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# Function to wait for and click the 'Show More Products' button
def click_show_more_button(driver):
    try:
        # Wait until the "Show More Products" button is clickable
        button = WebDriverWait(driver, 10).until(
            EC.element_to_be_clickable((By.XPATH, "//button[@aria-label='Show more products']"))
        )
        # Click the button using JavaScript
        driver.execute_script("arguments[0].click();", button)
        print("Button clicked using JavaScript.")
        time.sleep(2)  # Allow products to load
        return True
    except Exception as e:
        print("No more 'Show More Products' button or error:", e)
        return False
```

```python
# Initialize WebDriver and open the specified URL
url = "https://www.bestbuy.ca/en-ca/collection/laptops-on-sale/46082?path=category%253aComputers%2B%2526%2BTablets%253bcategory%253aLaptops%2B%2526%2BMacBooks%253bsoldandshippedby0enrchstring%253aBest%2BBuy"

driver = webdriver.Chrome() 
driver.get(url)  # Navigate to the specified URL
driver.set_window_size(1200, 900)  # Ensure all elements are visible
print(driver.title)  # Verify the page title
time.sleep(5)  # Allow time for the page to load
```

```python
# Try to click the 'Show More Products' button until all products are loaded
try:
    while True:
        if not click_show_more_button(driver):  # Call the function and check if it succeeds
            break

    print("All products loaded.")

    # Get the full HTML source after loading all products
    html = driver.page_source
    print("HTML source captured.")

finally:
    driver.quit()
    print("Driver has been quit.")
```

```python
# Parse the HTML using BeautifulSoup and extract the product information
html_soup = BeautifulSoup(html, "html.parser")
laptop_containers = html_soup.find_all("div", class_="style-module_col-xs-8__x3Zoo style-module_col-sm-12__ILoHz px-2")

names, prices, discounts, ratings, num_reviews = [], [], [], [], []

for container in laptop_containers:
    name = container.find("h4").text.strip() if container.find("h4") else "N/A"
    names.append(name)
    price = container.find("span", class_="screenReaderOnly_3anTj").text.strip() if container.find("span", class_="screenReaderOnly_3anTj") else "N/A"
    prices.append(price)
    discount = container.find("span", class_="discount-text").text.strip() if container.find("span", class_="discount-text") else "N/A"
    discounts.append(discount)
    ratings.append(container.get("data-rating", "N/A"))
    num_reviews.append(container.get("data-review-count", "N/A"))
```
