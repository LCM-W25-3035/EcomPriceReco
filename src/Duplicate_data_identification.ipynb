**IMPORT** **DATA**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#import the dataset
df1=pd.read_csv('/content/Merged file.csv')
df1
```

```python
df2=pd.read_csv('/content/Amazon Scraper Ecomm.csv')
df2
```

```python
df3=pd.read_csv('/content/Amazon Scraper Ecomm2.csv')
df3
```

**COMBINE DATASETS**

```python
#concatenate the datasets to compare across them
combined_df=pd.concat([df1,df2,df3])
combined_df
```

**IDENTICAL DATA **

```python
#check identical rows across all dataset
ident_rows = combined_df[combined_df.duplicated(keep=False)]
ident_rows
```

```python
#number of identical rows
num_ident_rows = ident_rows.shape[0]
num_ident_rows
```

Out of 23117 rows across the three datasets, only 370 rows are identical. This indicates that are relatively small subset of the data appears consistently across all three datasets, while majority of the row differs.
