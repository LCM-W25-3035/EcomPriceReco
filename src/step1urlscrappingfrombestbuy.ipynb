```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

# URL of the Best Buy Appliances category page
url = "https://www.bestbuy.ca/en-ca/category/appliances/26517?path=category%253AAppliances%253Bsoldandshippedby0enrchstring%253ABest%2BBuy"
```

```python
# Function to wait for and click the 'Show More Products' button
def click_show_more_button(driver):
    """
    Function to wait for and click the 'Show More Products' button using JavaScript.
    If the button is not found, the loop breaks.
    """
    try:
        # Wait until the "Show More Products" button is clickable
        button = WebDriverWait(driver, 10).until(
            EC.element_to_be_clickable((By.XPATH, "//button[@aria-label='Show more products']"))
        )
        # Click the button using JavaScript
        driver.execute_script("arguments[0].click();", button)
        print("Button clicked using JavaScript.")
        time.sleep(2)  # Allow products to load
        return True
    except Exception as e:
        print("No more 'Show More Products' button or error:", e)
        return False
```

```python
driver = webdriver.Chrome()  # Ensure ChromeDriver is installed and added to PATH
driver.get(url)  # Navigate to the specified URL
driver.set_window_size(1200, 900)  # Ensure all elements are visible
print(driver.title)  # Verify the page title
time.sleep(5)  # Allow time for the page to load
```

```python
# Function to capture product URLs and save them incrementally
def capture_product_urls(driver, collected_urls, output_file_path):
    """
    Capture product URLs from the current page and save them incrementally to a file.
    """
    product_elements = driver.find_elements(By.CSS_SELECTOR, "a[href*='/en-ca/product/']")
    new_urls = []

    # Extract and save only new URLs
    for product in product_elements:
        url = product.get_attribute("href")
        if url not in collected_urls:
            collected_urls.add(url)
            new_urls.append(url)

    # Write new URLs to the file
    if new_urls:
        with open(output_file_path, "a") as file:
            for url in new_urls:
                file.write(url + "\n")
        print(f"Captured {len(new_urls)} new product URLs and saved to file.")

```

```python
output_file_path='C:/Users/91892/Downloads/captured_product_urls.txt'
```

```python
# Main script
def main():
    # Set up the WebDriver
    driver = webdriver.Chrome()  # Ensure ChromeDriver is installed and added to PATH
    driver.get(url)  # Navigate to the specified URL
    driver.set_window_size(1200, 900)  # Ensure all elements are visible
    print(driver.title)  # Verify the page title
    time.sleep(5)  # Allow time for the page to load

    # Initialize a set to store collected product URLs (to avoid duplicates)
    collected_urls = set()

    try:
        # Continuously click the 'Show More Products' button and capture URLs
        while True:
            capture_product_urls(driver, collected_urls, output_file_path)  # Capture and save URLs
            if not click_show_more_button(driver):  # Click the 'Show More Products' button
                break

        print(f"All products loaded. URLs saved to {output_file_path}")

    finally:
        driver.quit()
        print("Driver has been quit.")
```

```python
if __name__ == "__main__":
    main()
```
