```python
# Load the dataset
file_path = "output_data2.csv"
data = pd.read_csv('/content/output_data2.csv')
```

```python
# Display the first few rows of the dataset
print("First few rows of the dataset:")
print(data.head())

# Basic information about the dataset
print("\nDataset Information:")
print(data.info())

# Summary statistics of numerical columns
print("\nSummary Statistics:")
print(data.describe())
```

```python
# ----------------------------
# Basic Statistics: Mean, Median, Mode
# ----------------------------
# Relevant numerical columns
columns_to_analyze = ['ratings', 'no_of_ratings', 'discount_price', 'actual_price']

print("\nBasic Statistics:")
for col in columns_to_analyze:
    print(f"\nColumn: {col}")
    # Convert the column to numeric, handling errors by coercing to NaN
    data[col] = pd.to_numeric(data[col], errors='coerce')
    print(f"Mean: {data[col].mean():.2f}")
    print(f"Median: {data[col].median():.2f}")

    # Calculate the mode and check if it's empty
    mode_result = data[col].mode()
    if not mode_result.empty:
        print(f"Mode: {mode_result[0]:.2f}")  # Access the first mode if available
    else:
        print(f"Mode: No mode found (empty or all unique values)")
```

```python
# ----------------------------
# 1. Ratings Distribution
# ----------------------------
print("\nAnalyzing Ratings Distribution...")
sns.histplot(data['ratings'], bins=10, kde=True)
plt.title('Distribution of Ratings')
plt.xlabel('Ratings')
plt.ylabel('Frequency')
plt.show()
```

```python
# ----------------------------
# 2. Top-Rated Products
# ----------------------------
print("\nTop 10 Products by Ratings:")
top_rated = data.sort_values(by='ratings', ascending=False).head(10)
print(top_rated[['name', 'ratings', 'no_of_ratings', 'discount_price']])
```

```python
# ----------------------------
# 3. Category-Level Insights
# ----------------------------
# Average Ratings by Main Category
print("\nAverage Ratings by Main Category:")
category_ratings = data.groupby('main_category')['ratings'].mean().sort_values(ascending=False)
print(category_ratings)

# Average Discount Price by Main Category
print("\nAverage Discount Price by Main Category:")
category_discount = data.groupby('main_category')['discount_price'].mean().sort_values(ascending=False)
print(category_discount)
```
